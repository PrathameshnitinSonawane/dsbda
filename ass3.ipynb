```python
#iris data set
import pandas as pd
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
cols = ['Sepal_Length','Sepal_Width','Petal_Length','Petal_Width','Species']
iris = pd.read_csv(url, names=cols)

# For each species:
print("Setosa:\n", iris[iris['Species']=='Iris-setosa'].describe())
print("Versicolor:\n", iris[iris['Species']=='Iris-versicolor'].describe())
print("Virginica:\n", iris[iris['Species']=='Iris-virginica'].describe())
```
