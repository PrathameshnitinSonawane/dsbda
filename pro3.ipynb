```python
import pandas as pd

# URL to the dataset
csv_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'

# Column names since the dataset has no headers
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

# Load dataset
df = pd.read_csv(csv_url, header=None, names=column_names)

print(df.head())  # Optional: to view data
```

```python
from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder
le = LabelEncoder()

# Apply label encoding to the 'species' column
df['species_encoded'] = le.fit_transform(df['species'])

# Display the first few rows
print(df[['species', 'species_encoded']].head())
```
