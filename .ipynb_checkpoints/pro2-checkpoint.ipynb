```python
 #1. Locate open Iris dataset from the URL csv_url ='https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
# 2. Perform the One Hot encoding, by considering Species as target variable.
```

```python
import pandas as pd

# Define the URL
csv_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'

# Define column names since the dataset does not include a header row
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

# Load the dataset
df = pd.read_csv(csv_url, header=None, names=column_names)

# Display first 5 rows
print(df.head())
```

```python
# One-hot encode the 'species' column
df_encoded = pd.get_dummies(df, columns=['species'])

# Display the first few rows of the new dataframe
print(df_encoded.head())
```
