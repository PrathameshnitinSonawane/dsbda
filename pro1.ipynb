```python
import pandas as pd

# Step 1: Load the Iris dataset from the URL
csv_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'

# Define column names as the dataset has no header
col_names = ['sepal.length', 'sepal.width', 'petal.length', 'petal.width', 'Species']
df = pd.read_csv(csv_url, names=col_names)

# Step 2a: Display total number of rows and columns
print("Total rows and columns:", df.shape)

# Step 2b: Display data types of each column
print("\nData types of each column:")
print(df.dtypes)

# Step 2c: Sort data by 'sepal.length' in descending order
sorted_df = df.sort_values(by='sepal.length', ascending=False)
print("\nTop rows after sorting by sepal.length (descending):")
print(sorted_df.head())

# Step 2d: Slice rows 11 to 20 and select only 'sepal.length' and 'Species' columns
sliced_df = df.loc[10:19, ['sepal.length', 'Species']]
print("\nSliced Data (Rows 11–20 with 'sepal.length' and 'Species'):")
print(sliced_df)
```
