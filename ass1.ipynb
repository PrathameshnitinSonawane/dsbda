```python
df['reading score'] = pd.to_numeric(df['reading score'], errors='coerce')#for cleeaning the data
df = df.dropna(subset=['reading score'])  # Optional: drop rows with NaNs
```

```python
#outliers using interquartile range(IQR)
sorted_rscore=sorted(df['reading score'])
sorted_rscore
q1=np.percentile(sorted_rscore,25)
q3=np.percentile(sorted_rscore,75)
print(q1,q3)
IQR=q3-q1
IQR
lwr_bound=q1-(1.5*IQR)
up_bound=q3+(1.5*IQR)
print(lwr_bound,up_bound)
r_outliers=[]
for i in sorted_rscore:
    if(i<lwr_bound or i> up_bound):
        r_outliers.append(i)
print(r_outliers)
```

```python
#Quantile based flooring and capping:
df2=pd.read_csv("demo1.csv")
df2
ninty_percentile=np.percentile(df2["math score"],90)
ninty_percentile
b=np.where(df2["math score"]>ninty_percentile,ninty_percentile,df2["math score"])
b
```

```python
df2.insert(1,"m score",b,True)
df2
```

```python
#Quantile based flooring and capping:
col=['reading score']
df2.boxplot(col)
```

```python
sorted_rscore=sorted(df2['reading score'])
median=np.median(sorted_rscore)
median
```

```python
refined_df2=df2
q1=np.percentile(sorted_rscore,25)
q3=np.percentile(sorted_rscore,75)
print(q1,q3)
IQR=q3-q1
IQR
lwr_bound=q1-(1.5*IQR)
upr_bound=q3+(1.5*IQR)
refined_df2['reading score']=np.where(refined_df2['reading score']>upr_bound,median,refined_df2['reading score'])
```
